```python
import psycopg2

# Database connection details
db_host = 'market-place-database.chakkagimn20.us-east-2.rds.amazonaws.com'
db_user = 'cmpeuser'
db_pass = 'CMPE_202'
db_name = 'marketplacedb'
db_port = 5432

try:
    conn = psycopg2.connect(
        host=db_host,
        user=db_user,
        password=db_pass,
        database=db_name,
        port=db_port
    )
    cursor = conn.cursor()
    print("Connection successful!")

    # ✅ Query to list all tables in the 'public' schema
    cursor.execute("""
        SELECT table_name
        FROM information_schema.tables
        WHERE table_schema = 'public'
        AND table_type = 'BASE TABLE';
    """)

    tables = cursor.fetchall()
    print("\n📋 Tables in your database:")
    for t in tables:
        print("-", t[0])

except Exception as e:
    print("❌ Error connecting to database:", e)

finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
```
